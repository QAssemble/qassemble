```python
def FimpconvertLocStc(mat : np.ndarray,flag, input_dict : dict):

    norb = mat.shape[0]
    ns = mat.shape[2]
    nprob = mat.shape[3]
    
    mat_ret = np.zeros((norb,norb,ns,nprob),dtype=complex,order='F')

    if flag == 1:
        
        for key, val in input_dict.items():
            nn = len(input_dict[key])
            iprob = int(key)-1
            orb1 = input_dict[key][0]
            for js in range(ns):
                print(orb1)
                for ii in range(0,nn):
                    orb2 = input_dict[key][ii]
                    print(orb2)
                    for iorb2 in orb2:
                        for jorb2 in orb2:
                            for iorb1 in orb1:
                                for jorb1 in orb1:
                                    mat_ret[iorb2,jorb2,js,iprob] = mat[iorb1,jorb1,js,iprob]
        return mat_ret
    
    elif flag == 0:
        for key, val in input_dict.items():
            iprob = int(key)-1
            for js in range(ns):
                orb1 = input_dict[key][0]
                print(orb1)
                for iorb1 in orb1:
                    for jorb1 in orb1:
                        mat_ret[iorb1,jorb1,js,iprob] = mat[iorb1,jorb1,js,iprob]
        
        return mat_ret
        

    
```
